# 实例分割应用实践

实例分割(Instance Segmentation)，用于在图片中标注出目标物体的所有示例，在自动驾驶、医学影像、高精度GIS识别、3D建模辅助等领域有广泛的应用。本例中，我们使用实例分割领域经典的Mask R-CNN模型进行案例学习。Mask R-CNN是由华人学者何凯明教授主导的一个多任务机器学习框架，能够在一个模型中完成图片中实例的分类和定位，并绘制出像素级的遮罩(Mask)。

## 实践内容

在本案例中，我们将会学习Mask R-CNN模型，用于图片的实例分割。读者将学习特征图、卷积神经网络、分类与回归任务等人工智能领域中常见的概念，并学习Mask R-CNN中RoI特征区域的生成和修正过程，以及像素级Mask的绘制方法。在学习过程中，我们会介绍COCO数据集，模型的训练方法，通过预训练模型和最小数据集，来训练自己的模型并进行图片的实例分割，让读者理解完整的模型训练过程。

## <a name="mask_rcnn_overview">1. Mask R-CNN模型概览</a>

Mask R-CNN模型是一个较为复杂的网络模型，包括三个主要的子网络：

- backbone网络
- RPN网络
- head网络

整体结构如下图所示：

![mrcnn_arch](./img/mrcnn_architecture.jpg)

下面，我们对这三个主要的网络进行一个简单的介绍

### backbone网络

Mask R-CNN的backbone网络，也称为骨干网，主要用于图像的特征提取。在Mask R-CNN之前，Faster R-CNN使用一个共享的卷积神经网络作为骨干网，Mask-RCNN的一个改进点在于，使用ResNet+FPN作为backbone网络，对于输入图片，生成多种尺寸的特征图，形成不同level的特征图金字塔，进一步强化了backbone网络的特征提取能力。

### RPN网络

RPN，即Region Proposal Network，用于生成待检测区域，通过一定的方法生成大量的Anchor(锚框)，在对Anchor进行进一步处理，区分前景、背景，计算Anchor区域与Ground Truth区域的偏移，对景背景分数和区域偏移量进行训练，生成RoI(Region of Interest)，即有可能框住物体的区域。RoI包含了前景背景分数以及位置偏移，结合Ground Truth信息，对RoI进一步的筛选和处理，将每个RoI赋上物体分类信息、Mask信息，再加上已有的位置偏移信息，得到Proposal，并传入后续的网络中进行处理。

### head网络

在RPN中生成的Proposal进入head网络之前，需要经过一个**RoI Align**层进行处理。在RoI向特征图进行映射时，可能出现像素值无法取整的情况，RoI Align层采用双线性插值算法计算出一个中心点的像素值，作为映射后的像素值。从而避免了Faster R-CNN中RoI Pooling取整后精度丢失的问题。

经过RoI Align后，head网络就可以结合Ground Truth box对RoI进行训练，head网络包括3个分支：

- RoI的分类
- RoI bounding box的回归
- Mask遮罩的回归

经过三个分支的训练，head网络就可以预测RoI的类别、包围框和像素级遮罩，最终完成实例分割的任务。
